## Creating SQL database 

```python
import sqlite3

conn = sqlite3.connect("../sql/bike_sharing.db")
```

```python
import pandas as pd
data_days = pd.read_csv("../data/processed/data_days.csv")
```

```python
data_days.to_sql("daily_rentals", conn, if_exists="replace", index=False)
```

```python
conn.commit()
```

## Initial queries

```python
query = """
SELECT *
FROM daily_rentals
LIMIT 5;
"""

pd.read_sql_query(query, conn)
```

```python
query = """ 
SELECT weathersit, AVG(cnt) AS avg_rentals
FROM daily_rentals
GROUP BY weathersit
"""
pd.read_sql_query(query, conn)
```

```python
query = """ 
    SELECT yr AS year, AVG(cnt) AS avg_rentals
    FROM daily_rentals
    GROUP BY yr
"""
pd.read_sql_query(query, conn)
```
